```python
ONNX_SESSION_OPTS = ort.SessionOptions()
ONNX_SESSION_OPTS.graph_optimization_level = ort.GraphOptimizationLevel.ORT_ENABLE_ALL
```

```python
yolo_model = ort.InferenceSession(
    "../models/yolo11m.onnx",
    sess_options=ONNX_SESSION_OPTS,
    providers=["TensorrtExecutionProvider",
        "CUDAExecutionProvider"],
)
```

```python
input_name = yolo_model.get_inputs()[0].name
input_shape = yolo_model.get_inputs()[0].shape
input_type = yolo_model.get_inputs()[0].type
```
